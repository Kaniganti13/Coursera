```python
import yfinance as yf
import pandas as pd
import plotly.express as px
```

```python
gme = yf.Ticker("GME")
gme_data = gme.history(period="max")
```

```python
gme_data.reset_index(inplace=True)
```

```python
def make_graph(data, title):
    fig = px.line(
        data,
        x="Date",
        y="Close",
        title=title
    )
    fig.show()
```

```python
make_graph(gme_data, "GameStop Stock Data")
```
